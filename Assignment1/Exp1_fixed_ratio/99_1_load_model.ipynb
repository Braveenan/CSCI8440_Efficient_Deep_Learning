```python
from vgg_cifar import vgg13
```

```python
model = vgg13()
```

```python
total_conv_weights = 0
total_filters = 0
```

```python
model
```

```python
for name, weight in model.named_parameters():

    # Conv weights have shape [N, C, Kh, Kw]
    if weight.dim() == 4:

        num_weights = weight.numel()
        num_filters = weight.shape[0]

        print(
            name,
            "shape =", tuple(weight.shape),
            "weights =", num_weights,
            "filters =", num_filters
        )

        total_conv_weights += num_weights
        total_filters += num_filters
```

```python
print("\nTotal Conv weights:", total_conv_weights)
print("Total Conv filters:", total_filters)
```
