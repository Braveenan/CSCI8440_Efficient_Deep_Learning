```python
import torch
```

```python
def unstructured_prune_test(tensor, sparsity):
    num_weights = tensor.numel()
    num_prune = int(num_weights * sparsity)

    if num_prune == 0:
        return torch.ones_like(tensor)

    if num_prune >= num_weights:
        mask = torch.zeros_like(tensor)
        tensor.mul_(mask)
        return mask

    magnitude = tensor.abs().flatten()
    threshold = torch.kthvalue(magnitude, num_prune).values

    mask = (tensor.abs() > threshold).to(tensor.dtype)
    tensor.mul_(mask)

    return mask
```

```python
w = torch.tensor([[0.8, -0.1], [0.5, -0.2]])

mask = unstructured_prune_test(w, 0.5)

print("Mask:", mask)
print("Pruned weights:", w)
```
